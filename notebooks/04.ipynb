```python
import sys;sys.path.append("..")
import pandas as pd, numpy as np
from src.features import build_features,FEATURE_SETS
daily = pd.read_csv("../data/processed/delhi_pollution_weather_daily_2020_2025.csv",
parse_dates=["Date"])
data = build_features(daily)
data.to_csv("../data/processed/delhi_features_daily.csv", index=False)
#--- leakage test: destroying the future must not change any past feature--
all_feats = sorted(set(sum(FEATURE_SETS.values(), [])))
num_cols = daily.select_dtypes("number").columns
t = 900
corrupt = daily.copy()
corrupt.loc[corrupt.index > t, num_cols] = 9999.0
a = build_features(daily).loc[:t, all_feats]
b = build_features(corrupt).loc[:t, all_feats]
assert a.equals(b), "LEAKAGE: a feature at time t depends on data after t"
print("Leakage test passed.")
```
